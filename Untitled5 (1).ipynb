```python
Q1. Word Overlap Analyzer (Strings + Sets) [5 marks]
Write a program that asks the user to enter two separate sentences. Treat each sentence as a bag of lowercase
words (ignore case and extra surrounding whitespace) and use set operations to report:
a) The words that appear in BOTH sentences (intersection).
b) The words that appear ONLY in the first sentence, and ONLY in the second sentence.
c) The total number of distinct words across both sentences combined (union), printed as a count.
d) The final &quot;unique to each sentence&quot; result printed as a sorted list, using the sorted() function.
```

```python
# Q1 
# take user input
sentence1=input("Enter first sentence: ").strip().lower()
sentence2=input("Enter second sentence: ").strip().lower()
# set() removes duplication 
words1=set(sentence1.split())
words2=set(sentence2.split())
# & finds words that are present in both sentences
both=words1 & words2
# - that are used for words that are only in the first set
only_first = words1 - words2
# - that are used for words that are that are only in the second set
only_second = words2 - words1
# | used for union
all_words=words1 | words2
unique_words = sorted(only_first | only_second)

print("\nWords in both sentences:", both)
print("Words only in first sentence:", only_first)
print("Words only in second sentence:", only_second)
print("Total distinct words:", len(all_words))
print("Unique to each sentence (sorted):", unique_words)
```

```python
Q2. Class Result Manager (Dictionaries + Sorting) [5 marks]
A dictionary stores student names as keys and a list of their marks (out of 100, multiple subjects) as values, for
example {&#39;Ali&#39;: [81, 90, 76]}. Write a program that:
a) Lets the user add a new student with a list of marks, and add an additional mark to an existing student.
b) Computes each student&#39;s average mark.
c) Prints the roster sorted by average mark in descending order, using sorted() with a custom key function
(not sort() on a plain list of numbers).
d) Clearly labels the student with the highest average as the &quot;Class Topper&quot;.
```

```python
# Q2 
students = {
    "Ali": [81, 90, 76],
    "Sara": [88, 92, 85],
    "Ahmed": [70, 75, 80]}
# Ask the user to take input
new_name=input("Enter new student name: ")
new_marks=input("Enter marks separated by spaces: ")
# Convert  marks into integers.
marks_list=[int(mark) for mark in new_marks.split()]
# Add the new student 
students[new_name]=marks_list
#  an existing student
existing_name=input("Enter student name to add an additional mark: ")
additional_mark=int(input("Enter additional mark: "))
if existing_name in students:
    students[existing_name].append(additional_mark)
else:
    print("Student not found.")

def calculate_average(student):
    name, marks = student
    return sum(marks) / len(marks)
# reverse=True means highest average comes first
sorted_students=sorted(
    students.items(),
    key=calculate_average,
    reverse=True   )
print("\nClass Roster:")
for index, (name, marks) in enumerate(sorted_students):
    average = sum(marks) / len(marks)
    if index == 0:
        print(f"{name}: {average:.2f} - Class Topper")
    else:
        print(f"{name}: {average:.2f}")
```

```python
Q3. Inventory Valuation (Tuples + Lists) [5 marks]
You are given inventory records as a list of tuples in the form (item_name, unit_price, quantity), e.g. [(&#39;Pen&#39;, 10,
50), (&#39;Notebook&#39;, 40, 20)]. Since each record is a tuple it cannot be modified in place. Write a program that:
a) Computes the total value (price × quantity) for each item without mutating the original tuples.
b) Builds a NEW list of tuples in the form (item_name, total_value) using the values computed above.
c) Finds and prints the single most expensive item by total value, and the grand total value of the entire
inventory.
d) Prints the new list sorted by total_value in descending order, using sorted() with key=lambda or a helper
function.
```

```python
# Q3 

inventory=[
    ("Pen", 10, 50),
    ("Notebook", 40, 20),
    ("Bag", 1500, 5),
    ("Pencil", 20, 30)    ]
total_values=[]
for item_name, unit_price, quantity in inventory:
    total_value=unit_price * quantity
    total_values.append((item_name, total_value))
most_expensive = max(
    total_values,
    key=lambda item: item[1]   )

# Add all 
grand_total=sum(
    value for name, value in total_values  )

#  from highest value to lowest value
sorted_inventory=sorted(
    total_values,
    key=lambda item: item[1],
    reverse=True   )
print("New inventory list:")
print(total_values)
print("\nMost expensive item:")
print(most_expensive)
print("\nGrand total value:")
print(grand_total)
print("\nInventory sorted by total value:")
print(sorted_inventory)
```

```python
Q4. Grade Classifier (if-elif-else + Dictionaries + Strings) [5 marks]
Write a program that stores student names and marks (out of 100) in a dictionary. For every student, classify their
grade using the following if-elif-else ladder: A (&gt;=85), B (75-84), C (60-74), D (50-59), F (below 50). The
program should:
a) Print each student&#39;s name and grade using str.format() or an f-string.
b) Maintain a second dictionary that counts how many students received each grade (e.g. {&#39;A&#39;: 2, &#39;B&#39;: 3, ...}).
c) After classifying everyone, print the grade-count summary sorted alphabetically by grade letter.
d) Handle invalid marks gracefully — if a mark is outside the 0-100 range, print an error message for that
student instead of crashing, and skip it in the grade count.
```

```python
# Q4 

students = {
    "Ali": 90,
    "Sara": 82,
    "Ahmed": 68,
    "Hassan": 55,
    "Ayesha": 45,
    "Usman": 105     }

# This dictionary count students 
grade_counts={
    "A": 0,
    "B": 0,
    "C": 0,
    "D": 0,
    "F": 0     }
# Check every student's marks
for name, marks in students.items():
    if marks < 0 or marks > 100:
        print(f"{name}: Invalid marks ({marks})")
        continue

    # Decide the grade 
    if marks >= 85:
        grade = "A"
    elif marks >= 75:
        grade = "B"
    elif marks >= 60:
        grade = "C"
    elif marks >= 50:
        grade = "D"
    else:
        grade = "F"
    # Print 
    print(f"{name}: Grade {grade}")
    grade_counts[grade] += 1
print("\nGrade Count Summary:")
# sorted() prints grades alphabetically.
for grade in sorted(grade_counts):
    print(f"{grade}: {grade_counts[grade]}")
```
